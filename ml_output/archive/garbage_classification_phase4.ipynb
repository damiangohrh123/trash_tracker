```python
from roboflow import Roboflow
from ultralytics import YOLO
import os

# Data download
rf = Roboflow(api_key="xPUpPw3Cakcrlvp5XMMl")
project = rf.workspace("damians-workspace-enpcl").project("garbage-classification-3-goezd")
version = project.version(4) 
dataset = version.download("yolo26")

# Point to the Phase 4 data.yaml
my_phase4_data = os.path.join(dataset.location, "data.yaml")
print(f"Phase 4 data is ready at: {my_phase4_data}")

# Load Model
model = YOLO('yolo26n.pt')

# Phase 4 Training
model.train(
    data=my_phase4_data,
    epochs=150,
    imgsz=800,               
    batch=16,
    device=0,
    workers=0,               
    name='trash_tracker_PHASE4',
    
    # YOLO26 Specific Optimizations
    optimizer='MuSGD',
    lr0=0.001,               
    cos_lr=True,             
    
    # Augmentations
    mosaic=1.0,              
    mixup=0.1,               
    degrees=15.0,            
    flipud=0.5,              

    save_period=5,           
    patience=30
)
```
