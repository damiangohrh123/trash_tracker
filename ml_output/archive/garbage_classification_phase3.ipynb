```python
from roboflow import Roboflow
from ultralytics import YOLO
import os

# Data Download (Version 3)
rf = Roboflow(api_key="xPUpPw3Cakcrlvp5XMMl")
project = rf.workspace("damians-workspace-enpcl").project("garbage-classification-3-goezd")
version = project.version(3) 
dataset = version.download("yolo26")

# Automatically point to the new data.yaml
my_phase3_data = os.path.join(dataset.location, "data.yaml")
print(f"Phase 3 data is ready at: {my_phase3_data}")

# Load YOLO26 Nano Model
model = YOLO('yolo26n.pt')

# Training
model.train(
    data=my_phase3_data,
    epochs=150,
    imgsz=640,
    batch=32,
    device=0,
    workers=0,                
    name='trash_tracker_PHASE3_FINAL',
    
    # AUGMENTATIONS
    mosaic=1.0,
    mixup=0.1,                
    hsv_h=0.015,              # Randomly shifts colors (helps distinguish paper/plastic)
    hsv_s=0.7,                
    hsv_v=0.4,                
    degrees=15.0,             
    translate=0.1,            
    scale=0.5,                
    flipud=0.5,               # Improves recognition of blurry/low-res Paper textures
    
    # OPTIMIZATION
    multi_scale=False,
    patience=50,
    optimizer='SGD',
    cos_lr=True,              # Smooths out learning spikes
    close_mosaic=20           # Turns off mosaic for the final 20 epochs for high precision
)
```
